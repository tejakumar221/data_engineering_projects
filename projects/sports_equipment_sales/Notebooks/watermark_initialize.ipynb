```python
%sql
use catalog end_to_end_projects;
use schema sports_equipments_sales;
```

```python
df_silver = spark.table("sports_equipment_sales_silver")
display(df_silver.limit(5))
print(f"Count Of Records in Silver Delta Table: {df_silver.count()}")
```

```python
%sql
select max(last_updated) as initial_watermark_value
from sports_equipment_sales_silver;
```

```python
%sql
create table if not exists watermark_control(
    source_name string,
    watermark_value timestamp
)
using delta;
```

```python
%sql
insert into watermark_control
select "sports_equipment" as source_name, 
max(last_updated) as watermark_value 
from sports_equipment_sales_silver;
```

```python
%sql
select * from watermark_control;
```
