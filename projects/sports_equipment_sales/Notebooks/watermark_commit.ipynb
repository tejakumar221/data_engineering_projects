```python
%sql
use catalog end_to_end_projects;
use schema sports_equipments_sales;
```

```python
processed_watermark = spark.sql("""
                                select max(last_updated)
                                from sports_equipment_sales_silver
                                """
).first()[0]

current_watermark = spark.sql("""
                             select watermark_value
                             from watermark_control
                             where source_name = "sports_equipment"
                             """
).first()[0]

print(f"Processed Watermark: {processed_watermark}")
print(f"Current Watermark: {current_watermark}")
                             
```

```python
spark.sql("""
          update watermark_control
          set watermark_value = :new_watermark
          where source_name = 'sports_equipment'
          and watermark_value < :new_watermark
          """, args = {"new_watermark" : processed_watermark}
)
```
